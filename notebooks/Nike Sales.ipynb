```python
def get_transaction_type(quantity):
    if pd.isna(quantity):
        return 'Unknown'
    elif quantity > 0:
        return 'Sale'
    elif quantity < 0:
        return 'Return'
    else:
        return 'No Sale'

df['transaction_type'] = df['units_sold'].apply(get_transaction_type)
```

```python
df[['units_sold', 'transaction_type']].head()
```

```python
df['revenue_clean'] = np.where(
    (df['units_sold'] > 0) & (df['mrp'].notna()),
    df['units_sold'] * df['mrp'] * (1 - df['discount_applied']),
    np.nan
)
```
